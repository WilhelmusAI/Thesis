````python
"""GitHub persistence for Colab runs. Tokens belong in Colab Secrets, never code."""

import base64
import hashlib
import random
import re
import time
from datetime import datetime, timezone
from email.utils import parsedate_to_datetime
from urllib.parse import quote

import requests


class GitHubStoreError(RuntimeError):
    """A GitHub persistence operation could not be completed safely."""


class GitHubConflictError(GitHubStoreError):
    """The requested immutable path already holds different content."""


class GitHubAPIError(GitHubStoreError):
    def __init__(self, message, status=None, retryable=False, retry_after=0):
        super().__init__(message)
        self.status = status
        self.retryable = retryable
        self.retry_after = retry_after


class GitHubStore:
    """Create immutable result files through GitHub's authenticated Contents API.

    Existing identical files are accepted; different files are never replaced.
    Requests use one fixed API host, and credentials are only sent in headers.
    A retry after an ambiguous upload first checks whether the upload succeeded.
    Calls are sequential: do not use one instance from concurrent upload threads.
    """

    API_ROOT = "https://api.github.com"
    MAX_UPLOAD_BYTES = 50 * 1024 * 1024
    MAX_ATTEMPTS = 5
    MAX_RETRY_SLEEP = 30.0

    def __init__(
        self, repo="WilhelmusAI/Thesis", branch="main", results_dir="Results",
        token=None, timeout=60,
    ):
        if not isinstance(repo, str) or not re.fullmatch(
            r"[A-Za-z0-9_.-]+/[A-Za-z0-9_.-]+", repo
        ):
            raise ValueError("repo must have the form 'owner/repository'.")
        if not isinstance(branch, str) or not branch.strip():
            raise ValueError("A nonempty GitHub branch is required.")
        if not isinstance(token, str) or not token.strip():
            raise ValueError("Load GITHUB_TOKEN from Colab Secrets before connecting.")
        if timeout <= 0:
            raise ValueError("timeout must be positive.")
        self.repo = repo
        self.branch = branch
        self.results_dir = self._clean_path(results_dir)
        self.timeout = timeout
        self._base = f"/repos/{repo}"
        self._session = requests.Session()
        self._session.headers.update({
            "Authorization": f"Bearer {token.strip()}",
            "Accept": "application/vnd.github.object+json",
            "X-GitHub-Api-Version": "2022-11-28",
            "User-Agent": "Thesis-Colab-Runner",
        })
        # Verify once at startup. Missing-path responses trigger re-verification.
        repository = self._json(self._request("GET", self._base))
        permissions = repository.get("permissions")
        if isinstance(permissions, dict) and permissions.get("push") is False:
            raise GitHubStoreError(
                f"The token cannot write to {repo}. Give the token Contents: read "
                "and write access to this repository."
            )
        self.resolve_ref()  # Fail early for an empty repository or wrong branch.

    @staticmethod
    def _clean_path(path, allow_empty=False):
        if not isinstance(path, str):
            raise TypeError("Repository paths must be strings.")
        if allow_empty and path == "":
            return ""
        if (
            not path or path.startswith("/") or "\\" in path
            or any(part in ("", ".", "..") for part in path.split("/"))
            or any(ord(character) < 32 for character in path)
        ):
            raise ValueError("Use a relative repository path without '..' or empty parts.")
        return path

    @staticmethod
    def blob_sha(data):
        """Git blob object ID; this is not the plain SHA-1 of the file bytes."""
        return hashlib.sha1(
            b"blob " + str(len(data)).encode("ascii") + b"\0" + data
        ).hexdigest()

    @staticmethod
    def _json(response):
        try:
            result = response.json()
        except (ValueError, TypeError):
            raise GitHubStoreError("GitHub returned an invalid JSON response.") from None
        if not isinstance(result, dict):
            raise GitHubStoreError("GitHub returned an unexpected response structure.")
        return result

    @staticmethod
    def _response_error(response):
        status = response.status_code
        headers = response.headers
        retry_after = 0.0
        value = headers.get("Retry-After", "")
        try:
            retry_after = max(0.0, float(value))
        except (ValueError, TypeError):
            try:
                retry_after = max(
                    0.0, (parsedate_to_datetime(value) - datetime.now(timezone.utc)).total_seconds()
                )
            except (ValueError, TypeError, OverflowError):
                pass
        limited = status == 429 or (
            status == 403 and (
                bool(value) or headers.get("X-RateLimit-Remaining") == "0"
            )
        )
        if status == 403 and not limited:
            # Inspect only for classification; never include response bodies in errors.
            try:
                message = str(response.json().get("message", "")).lower()
                limited = "rate limit" in message or "abuse detection" in message
            except (ValueError, TypeError, AttributeError):
                pass
        if limited and headers.get("X-RateLimit-Remaining") == "0":
            try:
                retry_after = max(retry_after, float(headers.get("X-RateLimit-Reset", 0)) - time.time())
            except (ValueError, TypeError):
                pass
        retryable = limited or status == 408 or 500 <= status < 600
        reasons = {
            401: "Authentication failed. Check the GITHUB_TOKEN secret.",
            403: "Access was denied. Check token Contents permissions and branch rules.",
            404: "The repository, branch, or resource is unavailable to this token.",
            409: "GitHub reported a repository conflict or an empty repository.",
            422: "GitHub rejected the write, possibly because the path already exists or a branch rule applies.",
        }
        reason = "GitHub rate limit reached." if limited else reasons.get(
            status, "GitHub did not complete the request."
        )
        return GitHubAPIError(
            f"GitHub HTTP {status}. {reason}", status, retryable, retry_after
        )

    def _pause(self, attempt, error):
        delay = min(
            self.MAX_RETRY_SLEEP,
            max(2 ** attempt + random.uniform(0, 0.5), error.retry_after),
        )
        time.sleep(delay)

    def _request(self, method, endpoint, *, allow_404=False, retry=True, **kwargs):
        """Retry reads safely; writes perform their own reconciliation before retry."""
        attempts = self.MAX_ATTEMPTS if retry else 1
        for attempt in range(attempts):
            try:
                response = self._session.request(
                    method, self.API_ROOT + endpoint, timeout=self.timeout,
                    allow_redirects=False, **kwargs,
                )
            except requests.RequestException:
                error = GitHubAPIError(
                    "A network error interrupted the GitHub request.", retryable=True
                )
            else:
                if 200 <= response.status_code < 300:
                    return response
                if allow_404 and response.status_code == 404:
                    return None
                if 300 <= response.status_code < 400:
                    raise GitHubStoreError(
                        "GitHub redirected the API request. Check whether the repository "
                        "was renamed; update the configured repository."
                    )
                error = self._response_error(response)
            if not error.retryable or attempt + 1 >= attempts:
                raise error from None
            self._pause(attempt, error)
        raise AssertionError("Unreachable retry state")

    def _commit_info(self):
        try:
            result = self._json(self._request(
                "GET", f"{self._base}/commits/{quote(self.branch, safe='')}",
                headers={"Accept": "application/vnd.github+json"},
            ))
        except GitHubAPIError as error:
            if error.status in (404, 409):
                raise GitHubStoreError(
                    f"Branch {self.branch!r} in {self.repo} is unavailable or empty. "
                    "Create the branch by committing the events JSON (or README) first, "
                    "and confirm the token can read it."
                ) from None
            raise
        if not re.fullmatch(r"[0-9a-f]{40}", str(result.get("sha", ""))):
            raise GitHubStoreError("GitHub did not return a valid commit SHA.")
        return result

    def resolve_ref(self):
        """Return the current branch's full commit SHA."""
        return self._commit_info()["sha"]

    def _verify_missing_path(self):
        # A private/deleted repo also gives 404; do not mistake it for a missing file.
        self._request("GET", self._base)
        self.resolve_ref()

    def _get_metadata_at(self, path, ref):
        path = self._clean_path(path)
        response = self._request(
            "GET", f"{self._base}/contents/{quote(path, safe='/')}",
            params={"ref": ref}, allow_404=True,
            headers={"Accept": "application/vnd.github.object+json"},
        )
        if response is None:
            self._verify_missing_path()
            return None
        return self._json(response)

    def get_metadata(self, path):
        """Return Contents metadata, or None only for a missing path in a valid repo."""
        return self._get_metadata_at(path, self.branch)

    def read_bytes(self, path, ref=None):
        """Read and blob-hash-verify a file, pinned to one commit throughout the read.

        Optional ref allows the caller to pin the event dataset to a known commit.
        Large files use the raw Contents media type, without following download URLs.
        """
        path = self._clean_path(path)
        pinned_ref = ref or self.resolve_ref()
        metadata = self._get_metadata_at(path, pinned_ref)
        if metadata is None:
            return None
        if metadata.get("type") != "file":
            raise GitHubStoreError(f"Expected a regular file at {path!r}.")
        response = self._request(
            "GET", f"{self._base}/contents/{quote(path, safe='/')}",
            params={"ref": pinned_ref},
            headers={"Accept": "application/vnd.github.raw+json"},
        )
        data = response.content
        if self.blob_sha(data) != metadata.get("sha"):
            raise GitHubStoreError(f"Git blob verification failed while reading {path!r}.")
        return data

    def _existing_identical(self, path, expected_sha):
        pinned_ref = self.resolve_ref()
        metadata = self._get_metadata_at(path, pinned_ref)
        if metadata is None:
            return None
        if metadata.get("type") == "file" and metadata.get("sha") == expected_sha:
            return pinned_ref
        raise GitHubConflictError(
            f"Refusing to overwrite {path!r}: it already contains different content. "
            "Use a different results folder for a different experiment."
        )

    def put_bytes_new(self, path, data, message):
        """Create a file and return a commit containing its verified bytes.

        Idempotent for byte-identical data. Never supplies GitHub's overwrite SHA.
        The conservative 50 MiB limit applies to decoded file bytes.
        """
        path = self._clean_path(path)
        if not isinstance(data, (bytes, bytearray)):
            raise TypeError("Upload data must be bytes or bytearray.")
        data = bytes(data)
        if len(data) > self.MAX_UPLOAD_BYTES:
            raise GitHubStoreError(
                f"{path!r} exceeds the 50 MiB upload limit. Reduce image dimensions "
                "or the amount of archived diagnostic data."
            )
        if not isinstance(message, str) or not message.strip():
            raise ValueError("A nonempty commit message is required.")
        expected_sha = self.blob_sha(data)
        existing_commit = self._existing_identical(path, expected_sha)
        if existing_commit:
            return existing_commit
        payload = {
            "message": message,
            "content": base64.b64encode(data).decode("ascii"),
            "branch": self.branch,
        }
        for attempt in range(self.MAX_ATTEMPTS):
            try:
                response = self._request(
                    "PUT", f"{self._base}/contents/{quote(path, safe='/')}",
                    json=payload, retry=False,
                    headers={"Accept": "application/vnd.github+json"},
                )
            except GitHubAPIError as error:
                if error.retryable or error.status in (409, 422):
                    # The server may have committed before the connection failed.
                    recovered_commit = self._existing_identical(path, expected_sha)
                    if recovered_commit:
                        return recovered_commit
                if not error.retryable or attempt + 1 >= self.MAX_ATTEMPTS:
                    raise
                self._pause(attempt, error)
                continue
            result = self._json(response)
            commit_sha = result.get("commit", {}).get("sha", "")
            content_sha = result.get("content", {}).get("sha")
            if not re.fullmatch(r"[0-9a-f]{40}", str(commit_sha)) or content_sha != expected_sha:
                raise GitHubStoreError(
                    f"GitHub returned an inconsistent upload acknowledgement for {path!r}. "
                    "Keep the local result; rerunning safely checks the remote file."
                )
            verified = self._get_metadata_at(path, commit_sha)
            if not verified or verified.get("sha") != expected_sha:
                raise GitHubStoreError(f"Remote blob verification failed for {path!r}.")
            return commit_sha
        raise AssertionError("Unreachable upload state")

    def _tree(self, tree_sha):
        result = self._json(self._request(
            "GET", f"{self._base}/git/trees/{tree_sha}",
            headers={"Accept": "application/vnd.github+json"},
        ))
        if result.get("truncated"):
            raise GitHubStoreError(
                "GitHub truncated a nonrecursive tree listing. Split results into "
                "smaller folders before relying on automatic resumption."
            )
        entries = result.get("tree")
        if not isinstance(entries, list):
            raise GitHubStoreError("GitHub returned an invalid tree listing.")
        return entries

    def list_dir(self, path=""):
        """Return immediate children keyed by filename, without the 1,000-file limit.

        Nonrecursive Git Trees calls walk the directory at one pinned commit.
        Missing directories return {}; a file at the requested directory raises.
        """
        path = self._clean_path(path, allow_empty=True)
        commit = self._commit_info()
        try:
            tree_sha = commit["commit"]["tree"]["sha"]
        except (KeyError, TypeError):
            raise GitHubStoreError("GitHub did not return the commit's root tree.") from None
        entries = self._tree(tree_sha)
        for part in path.split("/") if path else []:
            match = next((item for item in entries if item.get("path") == part), None)
            if match is None:
                return {}
            if match.get("type") != "tree":
                raise GitHubStoreError(
                    f"{path!r} is not a directory. GitHub folders are created by "
                    "placing files inside them; remove a plain file named Results."
                )
            entries = self._tree(match["sha"])
        output = {}
        for item in entries:
            name = item["path"]
            kind = {"tree": "dir", "blob": "file", "commit": "submodule"}.get(
                item.get("type"), item.get("type")
            )
            if item.get("mode") == "120000":
                kind = "symlink"
            output[name] = {
                "name": name,
                "path": f"{path}/{name}" if path else name,
                "type": kind,
                "sha": item["sha"],
                "size": item.get("size"),
                "commit_sha": commit["sha"],
            }
        return output

    def list_results(self):
        return {
            name: metadata for name, metadata in self.list_dir(self.results_dir).items()
            if metadata["type"] == "file" and name.lower().endswith(".zip")
        }

    def path_for(self, run_id):
        """Build Results/<Model_event_iteration>.zip from a safe run identifier."""
        if not isinstance(run_id, str) or not re.fullmatch(
            r"[A-Za-z0-9][A-Za-z0-9._-]*", run_id
        ):
            raise ValueError("run_id must contain only letters, digits, dots, '_' and '-'.")
        if run_id.lower().endswith(".zip"):
            raise ValueError("Pass run_id without its .zip extension.")
        return f"{self.results_dir}/{run_id}.zip"

    def close(self):
        self._session.close()
````

````python
# CELL 5 — PREBUILT CUDA ENGINE, PINNED MODELS AND LOCAL SERVER
import tarfile
import tempfile
from pathlib import PurePosixPath

_BACKEND_METADATA = {}
_BACKEND_ENVIRONMENTS = {}


def _backend_hash(path):
    h = hashlib.sha256()
    with Path(path).open("rb") as handle:
        for block in iter(lambda: handle.read(8 * 1024**2), b""):
            h.update(block)
    return h.hexdigest()


def _check_prebuilt_host():
    if platform.system() != "Linux" or platform.machine().lower() not in ("x86_64", "amd64"):
        raise RuntimeError("This pinned CUDA package requires a Linux x86-64 NVIDIA GPU runtime.")
    libc = platform.libc_ver()[1]
    if not libc or tuple(int(x) for x in libc.split(".")[:2]) < (2, 38):
        raise RuntimeError("The prebuilt engine needs glibc 2.38 or newer. Use a current Colab GPU runtime.")
    gpus = gpu_snapshot()
    for gpu in gpus:
        version = tuple(int(x) for x in gpu["driver"].split(".")[:2])
        if version < (570, 26):
            raise RuntimeError(
                f"NVIDIA driver {gpu['driver']} is too old for this CUDA 12.8 package. "
                "Use a Colab GPU runtime with driver 570.26 or newer; do not replace the host driver here.")
    print("GPU detected: " + ", ".join(g["name"] for g in gpus), flush=True)
    return gpus


def _download_backend_archive(spec, folder):
    """Resume interrupted downloads; accept only the pinned size and SHA256."""
    import requests
    from tqdm.auto import tqdm
    folder = Path(folder)
    folder.mkdir(parents=True, exist_ok=True)
    target = folder / spec["filename"]
    part = target.with_suffix(target.suffix + ".part")
    if target.is_file():
        if target.stat().st_size == spec["size_bytes"] and _backend_hash(target) == spec["sha256"]:
            print(f"Using verified download: {target.name}", flush=True)
            return target
        target.unlink()
    url = f"https://github.com/ggml-org/llama.cpp/releases/download/{LLAMA_CPP_RELEASE}/{spec['filename']}"
    with requests.Session() as session:
        for attempt in range(1, 4):
            try:
                offset = part.stat().st_size if part.exists() else 0
                if offset > spec["size_bytes"]:
                    part.unlink()
                    offset = 0
                if offset < spec["size_bytes"]:
                    headers = {"Accept-Encoding": "identity"}
                    if offset:
                        headers["Range"] = f"bytes={offset}-"
                    with session.get(url, headers=headers, stream=True, timeout=(15, 60)) as response:
                        response.raise_for_status()
                        if response.status_code == 206:
                            expected = f"bytes {offset}-{spec['size_bytes'] - 1}/{spec['size_bytes']}"
                            if response.headers.get("Content-Range") != expected:
                                raise RuntimeError("Unexpected byte range in backend download.")
                        elif response.status_code == 200:
                            offset = 0  # The server can ignore Range: restart, never append a full file.
                        else:
                            raise RuntimeError(f"Unexpected download status {response.status_code}.")
                        with part.open("ab" if offset else "wb") as handle, tqdm(
                                total=spec["size_bytes"], initial=offset, unit="B", unit_scale=True,
                                desc=spec["filename"], leave=False) as progress:
                            for block in response.iter_content(chunk_size=4 * 1024**2):
                                if block:
                                    if handle.tell() + len(block) > spec["size_bytes"]:
                                        raise RuntimeError("Backend download exceeds its pinned size.")
                                    handle.write(block)
                                    progress.update(len(block))
                if part.stat().st_size != spec["size_bytes"]:
                    raise requests.ConnectionError("Backend download ended before the expected size.")
                print(f"Verifying {spec['filename']}...", flush=True)
                if _backend_hash(part) != spec["sha256"]:
                    part.unlink()
                    raise RuntimeError("Backend archive SHA256 mismatch. Nothing was extracted; rerun to download again.")
                part.replace(target)
                return target
            except requests.RequestException as exc:
                if attempt == 3:
                    raise RuntimeError(
                        "Backend download failed after three attempts. Rerun Cell 8 to resume the partial download."
                    ) from exc
                print(f"Download interrupted; retry {attempt}/2 in {2 * attempt} seconds.", flush=True)
                time.sleep(2 * attempt)


def _extract_backend_archive(archive, destination):
    """Extract verified official archives while blocking unsafe paths and special files."""
    destination = Path(destination).resolve()
    if not hasattr(tarfile, "data_filter"):
        raise RuntimeError("Safe archive extraction needs a current Python runtime. Reconnect using current Colab.")
    with tarfile.open(archive, "r:gz") as handle:
        members = handle.getmembers()
        for member in members:
            name = PurePosixPath(member.name)
            if name.is_absolute() or ".." in name.parts:
                raise RuntimeError("Unsafe path in backend archive.")
            if not (member.isfile() or member.isdir() or member.issym() or member.islnk()):
                raise RuntimeError("Unsupported special file in backend archive.")
        handle.extractall(destination, members=members, filter="data")


def _backend_inventory(root):
    """Hash extracted files so a damaged cache is rebuilt from verified archives."""
    from tqdm.auto import tqdm
    root = Path(root)
    result = {}
    paths = sorted(p for p in root.rglob("*") if p.is_file() and p.name != "installation.json")
    for path in tqdm(paths, desc="Verify extracted CUDA files", unit="file", leave=False):
        if not path.resolve().is_relative_to(root.resolve()):
            raise ValueError("A cached CUDA file points outside its installation directory.")
        result[path.relative_to(root).as_posix()] = {"size": path.stat().st_size,
                                                   "sha256": _backend_hash(path)}
    return result


def backend_environment(binary):
    key = str(Path(binary).resolve())
    if key not in _BACKEND_ENVIRONMENTS:
        raise RuntimeError("Prepare and verify the CUDA engine before starting the model server.")
    return _BACKEND_ENVIRONMENTS[key].copy()


def _probe_prebuilt(binary, environment, log_path):
    """Fail before model downloads if the executable, flags or CUDA backend are unusable."""
    results = {}
    commands = {"version": "--version", "help": "--help", "devices": "--list-devices"}
    with Path(log_path).open("w", encoding="utf-8") as log:
        for name, flag in commands.items():
            print(f"Checking CUDA engine: {name}...", flush=True)
            try:
                run = subprocess.run([str(binary), flag], env=environment, capture_output=True,
                                     text=True, errors="replace", timeout=60)
            except subprocess.TimeoutExpired as exc:
                raise RuntimeError(f"CUDA engine {name} check timed out. Log: {log_path}") from exc
            output = run.stdout + run.stderr
            log.write(f"\n{flag}\n{output}\n")
            log.flush()
            if run.returncode:
                raise RuntimeError(f"Prebuilt engine {flag} failed. Check GPU/driver compatibility.\n"
                                   + output[-3000:] + f"\nLog: {log_path}")
            results[name] = output
    if LLAMA_CPP_COMMIT[:7] not in results["version"] or LLAMA_CPP_RELEASE[1:] not in results["version"]:
        raise RuntimeError("Downloaded engine reports a different version from the pinned release.")
    required = ["--jinja", "--reasoning", "--reasoning-format", "--reasoning-budget", "--no-context-shift",
                "--no-cache-prompt", "--cache-ram", "--fit", "--flash-attn", "--parallel",
                "--cache-type-k", "--cache-type-v", "--batch-size", "--ubatch-size"]
    absent = [flag for flag in required if not re.search(re.escape(flag) + r"(?=[\s,=]|$)", results["help"])]
    if absent:
        raise RuntimeError("Prebuilt engine lacks required options: " + ", ".join(absent))
    if not re.search(r"^\s*CUDA\d+:\s+\S", results["devices"], re.MULTILINE):
        raise RuntimeError("The engine cannot see a CUDA GPU.\n" + results["devices"][-3000:])
    return {"version_output": results["version"].strip(), "devices_output": results["devices"].strip()}


def build_llama_backend(work_dir):
    """Legacy function name retained for the runner: download a prebuilt engine, never compile."""
    global _BACKEND_METADATA
    _BACKEND_METADATA = {}
    _BACKEND_ENVIRONMENTS.clear()
    _check_prebuilt_host()
    base = Path(work_dir) / "prebuilt_backend" / LLAMA_CPP_RELEASE
    base.mkdir(parents=True, exist_ok=True)
    runtime = base / BACKEND_PLATFORM
    identity = {"method": "official-prebuilt", "release": LLAMA_CPP_RELEASE,
                "source_commit": LLAMA_CPP_COMMIT, "platform": BACKEND_PLATFORM,
                "archives": copy.deepcopy(BACKEND_ARCHIVES)}
    stamp = runtime / "installation.json"
    installed = None
    if stamp.is_file():
        try:
            candidate = json.loads(stamp.read_text(encoding="utf-8"))
            binary_relative = PurePosixPath(candidate["binary"])
            cached_binary = runtime / binary_relative
            safe_binary = (not binary_relative.is_absolute() and ".." not in binary_relative.parts
                           and binary_relative.name == "llama-server"
                           and cached_binary.resolve().is_relative_to(runtime.resolve())
                           and cached_binary.is_file() and os.access(cached_binary, os.X_OK)
                           and binary_relative.as_posix() in candidate["inventory"])
            if (safe_binary and candidate["identity"] == identity
                    and candidate["inventory"] == _backend_inventory(runtime)):
                installed = candidate
        except (OSError, ValueError, KeyError, TypeError):
            pass
    if installed is None:
        # Both compressed archives plus extracted libraries need several GB of disk.
        if shutil.disk_usage(base).free < 5 * 1024**3:
            raise RuntimeError("At least 5 GiB of free disk space is needed for the CUDA engine.")
        print(f"Downloading prebuilt CUDA engine {LLAMA_CPP_RELEASE} (about 767 MB). No compilation.", flush=True)
        archives = [_download_backend_archive(spec, base / "downloads") for spec in BACKEND_ARCHIVES]
        with tempfile.TemporaryDirectory(prefix="extract_", dir=base) as tmp:
            staging = Path(tmp) / "runtime"
            staging.mkdir()
            for archive in archives:
                print(f"Extracting {archive.name}...", flush=True)
                _extract_backend_archive(archive, staging)
            binaries = [p for p in staging.rglob("llama-server") if p.is_file()]
            if len(binaries) != 1:
                raise RuntimeError("Expected exactly one llama-server executable in the verified release.")
            binary = binaries[0]
            binary.chmod(binary.stat().st_mode | 0o111)
            installed = {"identity": identity, "binary": binary.relative_to(staging).as_posix(),
                         "inventory": _backend_inventory(staging)}
            atomic_json(staging / "installation.json", installed)
            if runtime.exists():
                shutil.rmtree(runtime)  # Only this release's failed/outdated extracted cache.
            staging.replace(runtime)
    else:
        print(f"Reusing verified CUDA engine {LLAMA_CPP_RELEASE}.", flush=True)
    binary = runtime / installed["binary"]
    library_dirs = sorted({str(p.parent.resolve()) for p in runtime.rglob("*.so*") if p.is_file()})
    environment = os.environ.copy()
    environment["LD_LIBRARY_PATH"] = os.pathsep.join(
        [str(binary.parent.resolve()), *library_dirs] +
        ([environment["LD_LIBRARY_PATH"]] if environment.get("LD_LIBRARY_PATH") else []))
    environment["CUDA_MODULE_LOADING"] = "LAZY"
    probes = _probe_prebuilt(binary, environment, base / "backend_check.log")
    _BACKEND_ENVIRONMENTS[str(binary.resolve())] = environment
    _BACKEND_METADATA = {**identity, **probes, "binary_sha256": _backend_hash(binary),
                         "verified_files": copy.deepcopy(installed["inventory"])}
    print("Prebuilt CUDA engine ready. Continuing to model download/loading.", flush=True)
    return binary




def download_model(model_id, hf_token=None):
    from huggingface_hub import hf_hub_download
    from tqdm.auto import tqdm
    spec = MODELS[model_id]
    free_gib = sum(g["free_mib"] for g in gpu_snapshot()) / 1024
    if free_gib < MIN_FREE_VRAM_GIB[model_id]:
        raise RuntimeError(
            f"{spec['name']}: {free_gib:.1f} GiB GPU memory free; the configured guard is "
            f"{MIN_FREE_VRAM_GIB[model_id]} GiB. Select a larger GPU. "
            "The script will not silently reduce context or change quantization.")
    folder = MODEL_CACHE_DIR / spec["name"] / spec["revision"]
    folder.mkdir(parents=True, exist_ok=True)
    target = folder / spec["filename"]
    if not target.exists() and shutil.disk_usage(folder).free < spec["size_bytes"] + 5 * 1024**3:
        raise RuntimeError(f"Insufficient disk space to download {spec['name']}.")
    target = Path(hf_hub_download(repo_id=spec["repo_id"], filename=spec["filename"],
                                revision=spec["revision"], local_dir=str(folder), token=hf_token))
    if target.stat().st_size != spec["size_bytes"]:
        raise RuntimeError(f"Downloaded GGUF size mismatch for {spec['name']}.")
    h = hashlib.sha256()
    with target.open("rb") as handle, tqdm(total=target.stat().st_size, unit="B",
            unit_scale=True, desc="Verify model SHA256", leave=False) as bar:
        for block in iter(lambda: handle.read(16 * 1024**2), b""):
            h.update(block)
            bar.update(len(block))
    if h.hexdigest() != spec["sha256"]:
        raise RuntimeError("Model SHA256 mismatch. Remove the corrupt download and retry.")
    return target


class LlamaServer:
    def __init__(self, binary, model_path, model_name, work_dir):
        import requests
        self.session = requests.Session()
        self.session.trust_env = False  # localhost does not need notebook proxy credentials.
        self.base_url = f"http://127.0.0.1:{SERVER_PORT}"
        self.process = None
        self.log_handle = None
        self.model_name = model_name
        self.log_path = Path(work_dir) / f"server_{model_name}.log"
        with socket.socket() as sock:
            if sock.connect_ex(("127.0.0.1", SERVER_PORT)) == 0:
                raise RuntimeError(f"Port {SERVER_PORT} is already in use; stop the old run first.")
        self.command = [str(binary), "--model", str(model_path), "--alias", model_name,
                        "--host", "127.0.0.1", "--port", str(SERVER_PORT),
                        "--ctx-size", str(CONTEXT_TOKENS), "--n-gpu-layers", str(GPU_LAYERS),
                        "--parallel", "1", "--jinja", "--reasoning", "on",
                        "--reasoning-format", "deepseek", "--no-context-shift",
                        "--reasoning-budget", str(THINKING_BUDGET_TOKENS),
                        "--no-cache-prompt", "--cache-ram", "0", "--fit", "off",
                        "--flash-attn", "on", "--batch-size", str(BATCH_SIZE),
                        "--ubatch-size", str(MICROBATCH_SIZE),
                        "--cache-type-k", "f16", "--cache-type-v", "f16",
                        "--timeout", str(REQUEST_TIMEOUT_SECONDS)]
        self.log_handle = self.log_path.open("w", encoding="utf-8")
        self.process = subprocess.Popen(self.command, stdout=self.log_handle,
                                        stderr=subprocess.STDOUT, env=backend_environment(binary))
        atexit.register(self.close)
        deadline = time.monotonic() + SERVER_START_TIMEOUT_SECONDS
        last_report = 0
        try:
            while time.monotonic() < deadline:
                if self.process.poll() is not None:
                    tail = self.log_path.read_text(encoding="utf-8", errors="replace")[-5000:]
                    raise RuntimeError("Model server exited during startup. Log tail:\n" + tail)
                try:
                    if self.session.get(self.base_url + "/health", timeout=3).status_code == 200:
                        self.props = self._post_or_get("GET", "/props")
                        actual = self.props.get("default_generation_settings", {}).get("n_ctx")
                        if actual is not None and actual < CONTEXT_TOKENS:
                            raise RuntimeError("Server context is smaller than configured; refusing truncation.")
                        print(f"{model_name} ready, thinking enabled.")
                        return
                except requests.RequestException:
                    pass
                if time.monotonic() - last_report > 20:
                    print(f"Loading {model_name}... startup log: {self.log_path}")
                    last_report = time.monotonic()
                time.sleep(2)
            raise TimeoutError("Server startup timed out; inspect " + str(self.log_path))
        except BaseException:
            self.close()
            raise

    def _post_or_get(self, method, endpoint, payload=None):
        started = time.perf_counter()
        response = self.session.request(method, self.base_url + endpoint, json=payload,
                                       timeout=(10, REQUEST_TIMEOUT_SECONDS))
        if response.status_code != 200:
            raise RuntimeError(f"Local inference server returned HTTP {response.status_code}: "
                               + response.text[:1200])
        result = response.json()
        if endpoint == "/v1/chat/completions":
            self._generation_request_seconds = time.perf_counter() - started
        return result

    def complete(self, messages, seed, sampling):
        payload = {"model": self.model_name, "messages": messages, "stream": False,
                   "seed": seed, "max_tokens": MAX_NEW_TOKENS, "cache_prompt": False,
                   "chat_template_kwargs": {"enable_thinking": True},
                   "reasoning_format": "deepseek", **sampling}
        template = self._post_or_get("POST", "/apply-template", payload)["prompt"]
        tokens = self._post_or_get("POST", "/tokenize", {
            "content": template, "add_special": True, "parse_special": True})["tokens"]
        if len(tokens) + MAX_NEW_TOKENS + 32 > CONTEXT_TOKENS:
            raise RuntimeError(f"Prompt requires {len(tokens)} tokens plus output budget; context exceeded. "
                               "Nothing was truncated. Revise protocol settings before collecting results.")
        started = time.monotonic()
        self._generation_request_seconds = None
        result = self._post_or_get("POST", "/v1/chat/completions", payload)
        choice = result["choices"][0]
        return {"raw_content": choice["message"].get("content") or "",
                "reasoning_content": choice["message"].get("reasoning_content") or "",
                "finish_reason": choice.get("finish_reason"), "usage": result.get("usage", {}),
                "prompt_tokens_checked": len(tokens), "duration_seconds": time.monotonic() - started,
                "generation_request_seconds": self._generation_request_seconds,
                "request": payload, "rendered_prompt": template, "raw_response": result}

    def close(self):
        if self.process and self.process.poll() is None:
            self.process.terminate()
            try:
                self.process.wait(timeout=15)
            except subprocess.TimeoutExpired:
                self.process.kill()
                self.process.wait(timeout=15)
        if self.log_handle and not self.log_handle.closed:
            self.log_handle.close()
        self.session.close()


# LIVE GENERATION MONITOR
# Run this cell after Cell 5 and before Cell 8.
# If Cell 8 is running, stop it first; do not restart/delete the runtime.
import threading
import subprocess
import time
import requests
from tqdm.auto import tqdm

GENERATION_PROGRESS_SECONDS = 10


def _thesis_live_request(self, method, endpoint, payload=None):
    original = type(self)._thesis_request_without_monitor
    if endpoint != "/v1/chat/completions":
        return original(self, method, endpoint, payload)

    stopped = threading.Event()
    started = time.monotonic()
    progress = tqdm(total=None, desc="Thinking + answer", unit="token",
                    position=4, leave=False, mininterval=0.5)
    progress.set_postfix_str("Sending request; waiting for first token")

    def monitor():
        previous = 0
        last_change = started
        with requests.Session() as session:
            session.trust_env = False
            while not stopped.wait(GENERATION_PROGRESS_SECONDS):
                status = "Token status unavailable"
                count = None
                try:
                    with session.get(self.base_url + "/slots", timeout=(1, 3)) as response:
                        response.raise_for_status()
                        slots = response.json()
                    active = [s for s in slots if isinstance(s, dict) and s.get("is_processing")]
                    if active:
                        value = active[0].get("next_token", {}).get("n_decoded")
                        if isinstance(value, int) and value >= 0:
                            count = value
                            status = "Generating" if count else "Preparing prompt / GPU kernels"
                        else:
                            status = "Request active; token count unavailable"
                    else:
                        status = "Waiting for server response"
                except (requests.RequestException, ValueError, TypeError, AttributeError):
                    status = "Server monitor has not responded"

                try:
                    gpu = subprocess.run(
                        ["nvidia-smi", "--query-gpu=utilization.gpu,memory.used,memory.total",
                         "--format=csv,noheader,nounits"],
                        capture_output=True, text=True, timeout=2)
                    fields = gpu.stdout.strip().splitlines()[0].split(",") if gpu.returncode == 0 else []
                    if len(fields) == 3:
                        used, total = float(fields[1]) / 1024, float(fields[2]) / 1024
                        status += f"; GPU {fields[0].strip()}%; VRAM {used:.1f}/{total:.1f} GiB"
                except (OSError, subprocess.TimeoutExpired, ValueError, IndexError):
                    pass

                if stopped.is_set():
                    break
                now = time.monotonic()
                if count is not None:
                    if count > previous:
                        progress.update(count - previous)
                        last_change = now
                        previous = count
                    elif now - last_change >= 30:
                        status += f"; no new tokens observed for {int(now - last_change)}s"
                progress.set_postfix_str(status)
                progress.refresh()

    worker = threading.Thread(target=monitor, daemon=True)
    worker.start()
    try:
        result = original(self, method, endpoint, payload)
        elapsed = time.monotonic() - started
        count = result.get("usage", {}).get("completion_tokens")
        label = f"{count:,} output tokens" if isinstance(count, int) else "response received"
        tqdm.write(f"Generation finished: {label} in {elapsed:.1f}s.")
        return result
    finally:
        stopped.set()
        worker.join(timeout=8)
        progress.close()


if "LlamaServer" not in globals():
    raise RuntimeError("Run Cell 5 first, then run this progress cell.")
if not hasattr(LlamaServer, "_thesis_request_without_monitor"):
    LlamaServer._thesis_request_without_monitor = LlamaServer._post_or_get
LlamaServer._post_or_get = _thesis_live_request
print("Live generation monitor enabled. Run Cell 8; cached downloads will be reused.")
````

````python
# Run performance instrumentation. Uses the Python standard library only.
import csv
import threading
import uuid

PERFORMANCE_SCHEMA_VERSION = 1


def _perf_number(value):
    if isinstance(value, bool):
        return None
    try:
        number = float(value)
        return number if math.isfinite(number) and number >= 0 else None
    except (ValueError, TypeError):
        return None


def _perf_ratio(numerator, denominator):
    return numerator / denominator if denominator and denominator > 0 else None


def performance_resource_snapshot():
    """VM-wide readings, not Python-process-only RAM. Missing is never zero."""
    result = {"timestamp_utc": utc_now(), "gpus": [], "system_ram_used_mib": None,
              "system_ram_total_mib": None, "system_ram_percent": None, "errors": []}
    try:
        memory = {}
        for line in Path("/proc/meminfo").read_text().splitlines():
            key, value = line.split(":", 1)
            memory[key] = int(value.strip().split()[0])  # Linux kB means KiB.
        total, available = memory["MemTotal"] / 1024, memory["MemAvailable"] / 1024
        result.update(system_ram_used_mib=total - available, system_ram_total_mib=total,
                      system_ram_percent=100 * (total - available) / total)
    except (OSError, ValueError, KeyError, IndexError):
        result["errors"].append("system_ram_unavailable")
    try:
        query = subprocess.run([
            "nvidia-smi", "--query-gpu=index,uuid,name,driver_version,utilization.gpu,memory.used,memory.total",
            "--format=csv,noheader,nounits"], capture_output=True, text=True, timeout=1.5)
        if query.returncode:
            raise ValueError("GPU query failed")
        for row in csv.reader(query.stdout.strip().splitlines(), skipinitialspace=True):
            if len(row) != 7:
                raise ValueError("Unexpected GPU query format")
            index, device_id, name, driver, utilization, used, total = (v.strip() for v in row)
            used, total = _perf_number(used), _perf_number(total)
            result["gpus"].append({"index": index, "uuid": device_id, "name": name,
                "driver": driver, "utilization_percent": _perf_number(utilization),
                "memory_used_mib": used, "memory_total_mib": total,
                "memory_used_percent": 100 * used / total if used is not None and total else None})
        if not result["gpus"]:
            raise ValueError("No GPU readings")
    except (OSError, subprocess.TimeoutExpired, ValueError):
        result["errors"].append("gpu_query_unavailable")
    return result


class RunPerformanceMonitor:
    """A fresh segment per resume; snapshots are copied before checkpoint writes."""
    def __init__(self, state):
        self.state = state
        self.interval = float(PERFORMANCE_SAMPLE_SECONDS)
        self.started = time.perf_counter()
        self.stopped = threading.Event()
        self.lock = threading.Lock()
        self.samples = []
        sessions = state.get("metadata", {}).get("runtime_sessions", [])
        self.segment = {"segment_id": uuid.uuid4().hex, "schema_version": PERFORMANCE_SCHEMA_VERSION,
            "started_at": utc_now(), "closed": False, "elapsed_seconds": 0.0,
            "runtime_session_index": len(sessions) - 1 if sessions else None,
            "attempt_start_index": len(state["attempts"]), "sample_interval_seconds": self.interval,
            "samples": []}
        state.setdefault("performance_segments", []).append(self.segment)
        self.worker = threading.Thread(target=self._loop, daemon=True)

    def _sample(self):
        try:
            sample = performance_resource_snapshot()
        except Exception as exc:
            sample = {"timestamp_utc": utc_now(), "gpus": [],
                      "errors": ["monitor_error:" + type(exc).__name__]}
        sample["elapsed_seconds"] = time.perf_counter() - self.started
        with self.lock:
            self.samples.append(sample)

    def _loop(self):
        if self.stopped.wait(self.interval):
            return
        while not self.stopped.is_set():
            before = time.perf_counter()
            self._sample()
            delay = max(0.05, self.interval - (time.perf_counter() - before))
            if self.stopped.wait(delay):
                return

    def start(self):
        # Initial reading makes short runs observable; its overhead is part of the run.
        self._sample()
        self.worker.start()

    def sync(self):
        with self.lock:
            self.segment["samples"] = copy.deepcopy(self.samples)
        if not self.segment["closed"]:
            self.segment["elapsed_seconds"] = time.perf_counter() - self.started
        self.segment["last_checkpoint_at"] = utc_now()

    def complete(self, backend, record, messages, seed, sampling):
        record["performance_segment_id"] = self.segment["segment_id"]
        started = time.perf_counter()
        try:
            return backend.complete(messages, seed, sampling)
        finally:
            # Includes template/token preflight, inference and progress cleanup, even on Ctrl-C.
            record["prompt_call_seconds"] = time.perf_counter() - started

    def stop(self):
        ended = time.perf_counter()
        self.stopped.set()
        if self.worker.ident is not None:
            self.worker.join(timeout=2)
        self.segment.update(closed=True, ended_at=utc_now(), elapsed_seconds=ended - self.started)
        self.sync()
        # A query already in flight at termination can finish after the run boundary.
        self.segment["samples"] = [s for s in self.segment["samples"]
                                   if s["elapsed_seconds"] <= self.segment["elapsed_seconds"]]


def performance_attempt_rows(state):
    rows = []
    for index, attempt in enumerate(state.get("attempts", []), 1):
        usage = attempt.get("usage") or {}
        raw = attempt.get("raw_response") or {}
        timings = raw.get("timings") or {}
        count = _perf_number(usage.get("completion_tokens"))
        elapsed = _perf_number(attempt.get("prompt_call_seconds"))
        decode_n = _perf_number(timings.get("predicted_n"))
        decode_ms = _perf_number(timings.get("predicted_ms"))
        prefill_ms = _perf_number(timings.get("prompt_ms"))
        rows.append({"run_id": state["run_id"], "attempt_index": index,
            "prompt_id": attempt.get("prompt_id"), "layer": attempt.get("layer"),
            "attempt": attempt.get("attempt"), "status": attempt.get("status"),
            "segment_id": attempt.get("performance_segment_id"),
            "prompt_call_seconds": elapsed,
            "generation_request_seconds": _perf_number(attempt.get("generation_request_seconds")),
            "legacy_duration_seconds": _perf_number(attempt.get("duration_seconds")),
            "prompt_tokens": _perf_number(usage.get("prompt_tokens")), "output_tokens": count,
            "output_tokens_per_second_end_to_end": _perf_ratio(count, elapsed) if count is not None else None,
            "server_decode_tokens": decode_n,
            "server_decode_seconds": decode_ms / 1000 if decode_ms is not None else None,
            "server_decode_tokens_per_second": _perf_ratio(1000 * decode_n, decode_ms) if decode_n is not None else None,
            "server_prefill_tokens": _perf_number(timings.get("prompt_n")),
            "server_prefill_seconds": prefill_ms / 1000 if prefill_ms is not None else None})
    return rows


def _perf_timing_summary(rows):
    completed = [r for r in rows if r["status"] in ("accepted", "rejected")]
    measured = [r for r in completed if r["prompt_call_seconds"] is not None]
    paired = [r for r in measured if r["output_tokens"] is not None and r["prompt_call_seconds"] > 0]
    decoded = [r for r in completed if r["server_decode_tokens"] is not None and r["server_decode_seconds"] and r["server_decode_seconds"] > 0]
    prefilled = [r for r in completed if r["server_prefill_tokens"] is not None and r["server_prefill_seconds"] and r["server_prefill_seconds"] > 0]
    timed = [r for r in rows if r["prompt_call_seconds"] is not None]
    accepted_timed = sum(r["status"] == "accepted" for r in timed)
    total_time = sum(r["prompt_call_seconds"] for r in timed)
    token_rows = [r for r in completed if r["output_tokens"] is not None]
    return {"attempt_count": len(rows), "accepted_prompt_count": sum(r["status"] == "accepted" for r in rows),
        "rejected_attempt_count": sum(r["status"] == "rejected" for r in rows),
        "interrupted_attempt_count": sum(r["status"] == "interrupted" for r in rows),
        "completed_attempt_count": len(completed), "timed_attempt_count": len(timed),
        "completed_attempts_with_wall_time": len(measured), "attempts_with_output_tokens": len(token_rows),
        "attempts_with_end_to_end_rate": len(paired), "attempts_with_decode_timings": len(decoded),
        "attempts_with_prefill_timings": len(prefilled),
        "known_output_tokens": sum(r["output_tokens"] for r in token_rows) if token_rows else None,
        "observed_prompt_call_seconds": total_time if timed else None,
        "average_prompt_generation_seconds": _perf_ratio(sum(r["prompt_call_seconds"] for r in measured), len(measured)),
        "average_seconds_per_accepted_prompt_including_retries": _perf_ratio(total_time, accepted_timed),
        "average_output_tokens_per_second_end_to_end": _perf_ratio(sum(r["output_tokens"] for r in paired), sum(r["prompt_call_seconds"] for r in paired)),
        "average_decode_tokens_per_second": _perf_ratio(sum(r["server_decode_tokens"] for r in decoded), sum(r["server_decode_seconds"] for r in decoded)),
        "average_prefill_tokens_per_second": _perf_ratio(sum(r["server_prefill_tokens"] for r in prefilled), sum(r["server_prefill_seconds"] for r in prefilled)),
        "average_server_prefill_seconds": _perf_ratio(sum(r["server_prefill_seconds"] for r in prefilled), len(prefilled))}


def _perf_weighted_resource(segments, getter):
    """Hold a reading until the next sample, capped at 2 intervals + query timeout.

    Never bridge session gaps or missing readings. Coverage accompanies every mean.
    """
    area = coverage = 0.0
    values = []
    for segment in segments:
        samples = segment.get("samples", [])
        duration = _perf_number(segment.get("elapsed_seconds")) or 0.0
        max_gap = 2 * segment.get("sample_interval_seconds", 2) + 1.5
        for index, sample in enumerate(samples):
            value = _perf_number(getter(sample))
            if value is None:
                continue
            values.append(value)
            start = sample["elapsed_seconds"]
            end = samples[index + 1]["elapsed_seconds"] if index + 1 < len(samples) else duration
            weight = max(0.0, min(end, duration, start + max_gap) - start)
            area += value * weight
            coverage += weight
    return {"average": _perf_ratio(area, coverage), "peak_sampled": max(values) if values else None,
            "coverage_seconds": coverage, "valid_sample_count": len(values)}


def _perf_resources(segments):
    metrics = {}
    for field in ("system_ram_used_mib", "system_ram_percent"):
        metrics[field] = _perf_weighted_resource(segments, lambda s, field=field: s.get(field))
    devices = {}
    for segment in segments:
        for sample in segment.get("samples", []):
            for gpu in sample.get("gpus", []):
                devices[gpu["uuid"]] = {k: gpu[k] for k in ("uuid", "index", "name", "driver", "memory_total_mib")}
    gpu_stats = []
    for device_id, device in devices.items():
        def get_device(sample, field):
            return next((g.get(field) for g in sample.get("gpus", []) if g["uuid"] == device_id), None)
        gpu_stats.append({**device, **{field: _perf_weighted_resource(segments,
                         lambda s, field=field: get_device(s, field))
                         for field in ("utilization_percent", "memory_used_mib", "memory_used_percent")}})
    # Across devices, utilization is a mean, memory is a total. Colab normally has one GPU.
    def aggregate(sample, field, average=False):
        values = [_perf_number(g.get(field)) for g in sample.get("gpus", [])]
        if not values or any(v is None for v in values):
            return None
        return sum(values) / len(values) if average else sum(values)
    metrics["gpu_utilization_percent"] = _perf_weighted_resource(segments, lambda s: aggregate(s, "utilization_percent", True))
    metrics["gpu_memory_used_mib"] = _perf_weighted_resource(segments, lambda s: aggregate(s, "memory_used_mib"))
    return metrics, gpu_stats


PERFORMANCE_DEFINITIONS = {
    "average_decode_tokens_per_second": "Sum of llama.cpp predicted_n / sum of predicted_ms in seconds over completed attempts with both fields. Includes generated thinking and answer tokens; excludes prompt prefill. Not an arithmetic mean of rates.",
    "average_output_tokens_per_second_end_to_end": "Sum of reported completion_tokens / sum of prompt-call wall seconds for completed attempts with both fields. Includes preflight, prefill, decoding, HTTP and progress cleanup. Includes format-rejected responses.",
    "average_prompt_generation_seconds": "Mean prompt-call wall time over completed accepted and rejected attempts with recorded wall time. Does not hide format retries.",
    "average_seconds_per_accepted_prompt_including_retries": "All recorded prompt-call seconds, including interrupted attempts and retries, divided by the number of accepted attempts with recorded wall time.",
    "average_prefill_tokens_per_second": "Sum of server prompt_n / sum of server prompt_ms in seconds for paired valid server timings.",
    "active_run_seconds": "Sum of recorded generation-loop segments. Includes validation, checkpoint and monitoring overhead, excludes model setup/loading, export, GitHub upload and downtime between sessions. Hard-killed segments only retain time through their last checkpoint.",
    "resource_averages": "Time-weighted sampled readings during the generation loop, held to the next reading with a maximum gap of 2 sampling intervals + 1.5 seconds. Missing intervals are excluded; coverage and sample counts are reported. Peaks are sampled maxima.",
    "gpu_usage": "nvidia-smi device-wide utilization (percent of the device sampling window with a kernel executing), not FLOP efficiency. Overall utilization averages visible GPUs; memory sums them. Per-device results are also stored.",
    "ram_usage": "VM-wide Linux MemTotal minus MemAvailable from /proc/meminfo, in MiB. Includes the model server and other VM activity, not just the Python process. 1024 MiB = 1 GiB.",
    "coverage": "Unavailable values are null/blank, never fabricated zero readings. Old runs lack new monitoring; interrupted requests may lack output tokens. Inspect coverage counts and closed segments before comparing.",
    "hardware": "Compare models on matching hardware/software and monitoring settings, or stratify results. Same-token budgets do not imply the same compute cost. Mixed hardware is flagged and segment reports allow separation.",
}


def performance_report(state):
    segments = state.get("performance_segments", [])
    rows = performance_attempt_rows(state)
    summary = {"run_id": state["run_id"], "model": state.get("config", {}).get("model", {}).get("name"),
        "requires_clean_main_run": bool(state.get("metadata", {}).get("format_recovery", {}).get("requires_clean_main_run")),
        "event_number": state["event"]["number"], "iteration": state.get("iteration"),
        "thinking_budget_tokens": state.get("config", {}).get("thinking_budget_tokens"),
        "performance_schema_version": PERFORMANCE_SCHEMA_VERSION, **_perf_timing_summary(rows),
        "active_run_seconds": sum(s.get("elapsed_seconds", 0) for s in segments) if segments else None,
        "recorded_segment_count": len(segments), "unclosed_segment_count": sum(not s.get("closed") for s in segments),
        "resource_sample_count": sum(len(s.get("samples", [])) for s in segments),
        "attempts_missing_new_wall_time": sum(r["prompt_call_seconds"] is None for r in rows)}
    resources, devices = _perf_resources(segments)
    for field, stats in resources.items():
        for statistic, value in stats.items():
            summary[f"{field}_{statistic}"] = value
    sessions = state.get("metadata", {}).get("runtime_sessions", [])
    hardware = []
    for session in sessions:
        gpus = session.get("environment", {}).get("gpus", [])
        signature = [{k: g.get(k) for k in ("name", "total_mib", "driver")} for g in gpus]
        if signature and signature not in hardware:
            hardware.append(signature)
    summary["hardware_configurations_seen"] = len(hardware)
    summary["mixed_hardware"] = len(hardware) > 1
    summary["gpu_models"] = " | ".join(sorted({g.get("name", "unknown") for h in hardware for g in h}
                                             | {g["name"] for g in devices}))
    summary["sample_interval_seconds"] = " | ".join(str(v) for v in sorted({s["sample_interval_seconds"] for s in segments}))
    segment_reports = []
    for segment in segments:
        sid = segment["segment_id"]
        index = segment.get("runtime_session_index")
        environment = sessions[index].get("environment") if isinstance(index, int) and 0 <= index < len(sessions) else None
        segment_resources, segment_devices = _perf_resources([segment])
        segment_reports.append({**{k: v for k, v in segment.items() if k != "samples"},
            "environment": environment, "timing": _perf_timing_summary([r for r in rows if r["segment_id"] == sid]),
            "resources": segment_resources, "devices": segment_devices})
    return {"schema_version": PERFORMANCE_SCHEMA_VERSION, "summary": summary, "definitions": PERFORMANCE_DEFINITIONS,
            "resources": resources, "devices": devices, "segments": segment_reports}


def performance_sample_rows(state):
    rows = []
    for segment in state.get("performance_segments", []):
        for index, sample in enumerate(segment.get("samples", []), 1):
            base = {"run_id": state["run_id"], "segment_id": segment["segment_id"], "sample_index": index,
                "timestamp_utc": sample.get("timestamp_utc"), "elapsed_seconds": sample["elapsed_seconds"],
                **{k: sample.get(k) for k in ("system_ram_used_mib", "system_ram_total_mib", "system_ram_percent")},
                "monitor_errors": " | ".join(sample.get("errors", []))}
            # One row per GPU per sample. RAM is repeated; do not sum it across GPU rows.
            for gpu in sample.get("gpus", []) or [{}]:
                rows.append({**base, **{"gpu_" + k: gpu.get(k) for k in (
                    "uuid", "index", "name", "driver", "utilization_percent", "memory_used_mib", "memory_total_mib", "memory_used_percent")}})
    return rows


def archive_has_performance(path):
    # Historical function name retained for patches in already-running notebooks.
    # Rebuild a cached legacy export before its first upload, using the local checkpoint.
    try:
        with zipfile.ZipFile(path) as archive:
            return (archive.namelist() == ["results.xlsx"]
                    and archive.comment == b"thesis-excel-only-v1")
    except (OSError, KeyError, ValueError, zipfile.BadZipFile):
        return False


# CELL 6 - Export a complete run for inspection and manual evaluation.
# These functions run in Colab. They do not call or evaluate with another model.
from pathlib import Path
import hashlib
import io
import json
import math
import re
import tempfile
import zipfile

EXCEL_ARCHIVE_FORMAT = "thesis-excel-only-v1"


def _workbook_resume_payload(state):
    """Only immutable result fields; manual evaluation cells may be edited later."""
    node_keys = ("node_id", "parent_id", "depth", "child_index", "sentences", "text",
                 "prompt_id", "accepted_attempt", "seed")
    return {"run_id": state["run_id"], "run_fingerprint": state["run_fingerprint"],
            "status": state["status"], "iteration": state["iteration"],
            "event": state["event"], "config": state["config"],
            "nodes": [{k: n.get(k) for k in node_keys} for n in state["nodes"]]}


def _export_json(value):
    return json.dumps(value, ensure_ascii=False, indent=2, default=str)


def _export_public_state(value):
    """Defence in depth: credentials are never part of an exported run."""
    secret_keys = {
        "github_token", "gh_token", "hf_token", "huggingface_token",
        "access_token", "auth_token", "api_key", "authorization", "password",
        "secret", "token", "headers", "request_headers",
    }
    if isinstance(value, dict):
        return {str(k): _export_public_state(v) for k, v in value.items()
                if str(k).lower() not in secret_keys}
    if isinstance(value, (list, tuple)):
        return [_export_public_state(v) for v in value]
    if isinstance(value, Path):
        return str(value)
    return value


def _export_node_sort(node):
    return tuple(int(p) if p.isdigit() else p for p in str(node["node_id"]).split("."))


def _export_tree_parts(state):
    nodes = sorted(state.get("nodes", []), key=_export_node_sort)
    lookup = {str(n["node_id"]): n for n in nodes}
    if len(lookup) != len(nodes) or "root" in lookup:
        raise ValueError("Export requires unique node IDs; 'root' is reserved.")
    children = {"root": []}
    for n in nodes:
        node_id = str(n["node_id"])
        parent = str(n["parent_id"])
        if parent != "root" and parent not in lookup:
            raise ValueError(f"Node {node_id} has missing parent {parent}.")
        children.setdefault(parent, []).append(node_id)
        children.setdefault(node_id, [])
    leaves = [n for n in nodes if not children[n["node_id"]]]
    branches = []
    for leaf in leaves:
        chain, seen, node_id = [], set(), leaf["node_id"]
        while node_id != "root":
            if node_id in seen:
                raise ValueError("Cycle detected in the exported tree.")
            seen.add(node_id)
            node = lookup[node_id]
            chain.append(node)
            node_id = str(node["parent_id"])
        branches.append(list(reversed(chain)))
    if nodes and len({n["node_id"] for b in branches for n in b}) != len(nodes):
        raise ValueError("Export found nodes not connected to the root.")
    return nodes, lookup, children, leaves, branches


def _export_workbook(state, path, nodes, branches):
    from openpyxl import Workbook
    from openpyxl.styles import Alignment, Font, PatternFill, Border, Side
    from openpyxl.utils import get_column_letter
    from openpyxl.worksheet.datavalidation import DataValidation

    wb = Workbook()
    summary = wb.active
    summary.title = "Run summary"
    chunks = []
    xml_controls = re.compile(r"[\x00-\x08\x0b\x0c\x0e-\x1f]")
    navy, teal, pale, ink = "19384A", "1F746F", "EAF2F3", "233744"
    gray, blue = "647481", "1A62B0"

    def set_cell(ws, row, col, value):
        if isinstance(value, (dict, list, tuple)):
            value = _export_json(value)
        cell = ws.cell(row, col)
        if isinstance(value, str):
            needs_escape = bool(xml_controls.search(value))
            reserved_marker = (ws.title != "Text chunks"
                               and value.startswith("[Full text in Text chunks: "))
            if needs_escape or reserved_marker or len(value.encode("utf-16-le")) // 2 > 32000:
                source = f"{ws.title}!{get_column_letter(col)}{row}"
                encoding = "json_string" if needs_escape else "plain"
                stored = json.dumps(value, ensure_ascii=False) if needs_escape else value
                pieces = [stored[i:i + 15000] for i in range(0, len(stored), 15000)]
                for part, piece in enumerate(pieces, 1):
                    chunks.append((source, part, len(pieces), piece, encoding))
                value = f"[Full text in Text chunks: {source}; {len(pieces)} parts]"
            cell.value = value
            # Explicit string typing prevents generated text becoming an Excel formula.
            cell.data_type = "s"
            if value.lstrip().startswith(("=", "+", "-", "@")):
                cell.quotePrefix = True
        else:
            cell.value = value
        cell.alignment = Alignment(vertical="top", wrap_text=True)
        cell.font = Font(name="Calibri", size=11, color=ink)
        return cell

    def add_row(ws, values, row=None):
        row = row or ws.max_row + 1
        for col, value in enumerate(values, 1):
            set_cell(ws, row, col, value)
        return row

    def header(ws, labels, widths):
        for col, label in enumerate(labels, 1):
            cell = set_cell(ws, 1, col, label)
            cell.fill = PatternFill("solid", fgColor=navy)
            cell.font = Font(name="Calibri", size=11, bold=True, color="FFFFFF")
        ws.row_dimensions[1].height = 32
        for col, width in enumerate(widths, 1):
            ws.column_dimensions[get_column_letter(col)].width = width
        ws.freeze_panes = "A2"
        ws.sheet_view.showGridLines = False
        ws.sheet_properties.pageSetUpPr.fitToPage = True
        ws.page_setup.orientation = "landscape"
        ws.page_setup.paperSize = ws.PAPERSIZE_A3
        ws.page_setup.fitToWidth = 1
        ws.page_setup.fitToHeight = 0
        ws.print_title_rows = "1:1"

    def finish_table(ws, row_height=100, editable_columns=()):
        ws.auto_filter.ref = ws.dimensions
        thin = Side(style="hair", color="D8E3E7")
        for row in ws.iter_rows(min_row=2):
            ws.row_dimensions[row[0].row].height = row_height
            for cell in row:
                if cell.row % 2 == 0:
                    cell.fill = PatternFill("solid", fgColor="F4F8F9")
                cell.border = Border(bottom=thin)
                if cell.column in editable_columns:
                    cell.fill = PatternFill("solid", fgColor="EEF5FC")
                    cell.font = Font(name="Calibri", size=11, color=blue)

    def manual_validations(ws, outcome_col, plausibility_col):
        if ws.max_row < 2:
            return
        outcome = DataValidation(
            type="list", formula1='"Supported,Partially supported,Refuted,Not supported"',
            allow_blank=True,
        )
        outcome.errorTitle = "Choose one outcome"
        outcome.error = "Use a category from the dropdown or leave the field blank."
        outcome.errorStyle = "stop"
        outcome.showErrorMessage = True
        outcome.showInputMessage = True
        outcome.promptTitle = "Manual evaluation"
        outcome.prompt = "Use the definitions in your thesis. No automatic rating is supplied."
        ws.add_data_validation(outcome)
        outcome.add(f"{get_column_letter(outcome_col)}2:{get_column_letter(outcome_col)}{ws.max_row}")
        rating = DataValidation(type="whole", operator="between", formula1="1", formula2="5", allow_blank=True)
        rating.errorTitle = "Use an integer from 1 to 5"
        rating.error = "Enter 1, 2, 3, 4 or 5, or leave this field blank."
        rating.errorStyle = "stop"
        rating.showErrorMessage = True
        ws.add_data_validation(rating)
        rating.add(f"{get_column_letter(plausibility_col)}2:{get_column_letter(plausibility_col)}{ws.max_row}")

    event = state["event"]
    header(summary, ["Field", "Value"], [28, 116])
    summary_rows = [
        ("Run ID", state["run_id"]),
        ("Status", state.get("status", "unknown")),
        ("Run fingerprint", state.get("run_fingerprint", "")),
        ("Event number", event["number"]),
        ("Event title", event["Title"]),
        ("Original event description", event["description"]),
        ("Generated nodes", len(nodes)),
        ("Terminal branches", len(branches)),
        ("Configured layers", state.get("config", {}).get("num_layers", "")),
        ("Children per parent", state.get("config", {}).get("children_per_node", "")),
        ("Evaluation", "All manual evaluation fields are blank. Use the thesis definitions for Supported, Partially supported, Refuted and Not supported, and plausibility 1–5."),
        ("Archive contents", "This ZIP contains only results.xlsx. Predictions, branches, evaluation fields, generation logs, settings and performance measurements are included as worksheets."),
        ("Text preservation", "Strings are stored as text, not Excel formulas. Long text is split across Text chunks, with reversible encoding for XML control characters. Expand rows or use the formula bar for long worksheet text."),
    ]
    for item in summary_rows:
        add_row(summary, item)
    finish_table(summary, row_height=34)
    summary.row_dimensions[7].height = 300
    for r in (12, 13, 14):
        summary.row_dimensions[r].height = 52
    summary.auto_filter.ref = None

    nws = wb.create_sheet("Nodes")
    labels = ["Node ID", "Parent ID", "Depth", "Child index", "Sentence 1", "Sentence 2", "Sentence 3", "Sentence 4", "Sentence 5", "Complete consequence", "Outcome category", "Plausibility (1–5)", "Evidence / source", "Reviewer notes", "Prompt ID", "Accepted attempt", "Seed", "Generation order"]
    header(nws, labels, [16, 16, 9, 12] + [50] * 5 + [90, 24, 20, 48, 48, 22, 20, 16, 20])
    generation_order = {n["node_id"]: i for i, n in enumerate(state["nodes"], 1)}
    for n in nodes:
        sentences = list(n.get("sentences", []))
        if len(sentences) != 5:
            raise ValueError(f"Node {n['node_id']} must contain exactly five sentences before export.")
        add_row(nws, [n["node_id"], n["parent_id"], n["depth"], n.get("child_index", "")] + sentences + [n.get("text", " ".join(sentences)), None, None, None, None, n.get("prompt_id"), n.get("accepted_attempt"), n.get("seed"), generation_order[n["node_id"]]])
    finish_table(nws, 145, (11, 12, 13, 14))
    manual_validations(nws, 11, 12)

    bws = wb.create_sheet("Branches")
    depth = max([int(n["depth"]) for n in nodes] + [int(state.get("config", {}).get("num_layers", 0))])
    base = ["Branch", "Leaf node ID", "Node path", "Original event title", "Original event description"]
    labels = base + [f"Layer {d} consequence" for d in range(1, depth + 1)] + ["Outcome category", "Plausibility (1–5)", "Evidence / source", "Reviewer notes"]
    header(bws, labels, [10, 20, 40, 44, 96] + [90] * depth + [24, 20, 48, 48])
    for index, branch in enumerate(branches, 1):
        by_depth = {int(n["depth"]): n for n in branch}
        texts = [by_depth[d].get("text", " ".join(by_depth[d]["sentences"])) if d in by_depth else "" for d in range(1, depth + 1)]
        add_row(bws, [index, branch[-1]["node_id"], "root → " + " → ".join(n["node_id"] for n in branch), event["Title"], event["description"]] + texts + [None] * 4)
    finish_table(bws, 220, tuple(range(6 + depth, 10 + depth)))
    manual_validations(bws, 6 + depth, 7 + depth)

    aws = wb.create_sheet("Attempts")
    attempt_labels = ["Prompt ID", "Parent ID", "Attempt", "Seed", "Validation errors", "Raw final content", "Raw thinking content", "Full attempt record"]
    header(aws, attempt_labels, [24, 18, 12, 16, 60, 100, 100, 100])
    for attempt in state.get("attempts", []):
        add_row(aws, [attempt.get("prompt_id", ""), attempt.get("parent_id", ""), attempt.get("attempt", ""), attempt.get("seed", ""), attempt.get("validation_errors", []), attempt.get("raw_content", ""), attempt.get("reasoning_content", ""), attempt])
    finish_table(aws, 100)

    pws = wb.create_sheet("Performance")
    header(pws, ["Metric", "Value"], [64, 100])
    for key, value in state["performance"]["summary"].items():
        add_row(pws, [key, value])
    finish_table(pws, 30)
    pws.auto_filter.ref = None
    dws = wb.create_sheet("Metric definitions")
    header(dws, ["Metric / scope", "Definition"], [60, 120])
    for key, value in state["performance"]["definitions"].items():
        add_row(dws, [key, value])
    finish_table(dws, 70)
    for title, rows in (("Prompt metrics", performance_attempt_rows(state)),
                        ("Resource samples", performance_sample_rows(state)),
                        ("Performance segments", state["performance"]["segments"]),
                        ("GPU metrics", state["performance"]["devices"])):
        ws = wb.create_sheet(title)
        keys = list(dict.fromkeys(k for row in rows for k in row)) or ["run_id"]
        header(ws, keys, [48 if k in ("run_id", "segment_id", "gpu_uuid", "gpu_name") else 28 for k in keys])
        for row in rows:
            add_row(ws, [row.get(key) for key in keys])
        finish_table(ws, 30)

    mws = wb.create_sheet("Settings")
    header(mws, ["Section", "Setting", "Value", "Value encoding"], [20, 40, 100, 20])
    for section in ("config", "metadata"):
        for key, value in state.get(section, {}).items():
            encoding = "json" if isinstance(value, (dict, list, tuple)) else type(value).__name__
            add_row(mws, [section, key, value, encoding])
    finish_table(mws, 48)
    rws = wb.create_sheet("Archive metadata")
    header(rws, ["Field", "Value"], [30, 100])
    for key, value in {"archive_format": EXCEL_ARCHIVE_FORMAT,
                       "iteration": state["iteration"],
                       "resume_sha256": digest(_workbook_resume_payload(state)),
                       "performance_schema_version": PERFORMANCE_SCHEMA_VERSION}.items():
        add_row(rws, [key, value])
    finish_table(rws, 32)
    if chunks:
        cws = wb.create_sheet("Text chunks")
        header(cws, ["Source cell", "Part", "Total parts", "Text", "Encoding"], [30, 10, 12, 120, 20])
        # The list is fixed here; each 15,000-character part is below Excel's limit.
        for item in tuple(chunks):
            add_row(cws, item)
        finish_table(cws, 180)
    wb.save(path)


def export_run(state: dict, output_dir: Path) -> Path:
    """Atomically export only the workbook, including completion-validation data."""
    state = _export_public_state(state)
    state["performance"] = performance_report(state)
    run_id = str(state["run_id"])
    if not re.fullmatch(r"[A-Za-z0-9_.-]+", run_id) or run_id in (".", ".."):
        raise ValueError("run_id may contain only letters, numbers, underscores, dots and hyphens.")
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)
    nodes, lookup, children, leaves, branches = _export_tree_parts(state)
    final_zip = output_dir / f"{run_id}.zip"
    with tempfile.TemporaryDirectory(prefix=f"export_{run_id}_", dir=output_dir) as temp:
        directory = Path(temp)
        workbook_path = directory / "results.xlsx"
        _export_workbook(state, workbook_path, nodes, branches)
        zip_path = directory / f"{run_id}.zip"
        with zipfile.ZipFile(zip_path, "w", compression=zipfile.ZIP_DEFLATED, compresslevel=6) as archive:
            archive.comment = EXCEL_ARCHIVE_FORMAT.encode("ascii")
            archive.write(workbook_path, arcname="results.xlsx")
        zip_path.replace(final_zip)
    return final_zip


def read_completed_workbook(raw):
    """Read the workbook's existing tables; no hidden full-state copy is required."""
    from openpyxl import load_workbook
    with zipfile.ZipFile(io.BytesIO(raw)) as inner:
        if sum(e.file_size for e in inner.infolist()) > 300 * 1024**2:
            raise ValueError("Workbook exceeds the permitted expanded size.")
    wb = load_workbook(io.BytesIO(raw), read_only=True, data_only=False, keep_links=False)
    try:
        def table(name):
            sheet = wb[name]
            if sheet.max_row > 200000 or sheet.max_column > 128:
                raise ValueError(f"Worksheet {name} exceeds permitted dimensions.")
            return list(sheet.iter_rows())

        chunks = {}
        if "Text chunks" in wb.sheetnames:
            for row in table("Text chunks")[1:]:
                source, part, total, text, encoding = [c.value for c in row[:5]]
                if (not isinstance(source, str) or type(part) is not int or type(total) is not int
                        or not 1 <= part <= total <= 20000 or not isinstance(text, str)
                        or encoding not in ("plain", "json_string")):
                    raise ValueError("Invalid workbook text chunk.")
                group = chunks.setdefault(source, {"total": total, "encoding": encoding, "parts": {}})
                if group["total"] != total or group["encoding"] != encoding or part in group["parts"]:
                    raise ValueError("Duplicate or inconsistent workbook text chunk.")
                group["parts"][part] = text
        for group in chunks.values():
            if sorted(group["parts"]) != list(range(1, group["total"] + 1)):
                raise ValueError("Missing workbook text chunk.")
            value = "".join(group["parts"][i] for i in range(1, group["total"] + 1))
            group["value"] = json.loads(value) if group["encoding"] == "json_string" else value
            if not isinstance(group["value"], str):
                raise ValueError("Text chunk did not decode to text.")

        def read(cell, sheet):
            if cell.value is None:
                return None
            if cell.data_type == "f":
                raise ValueError("A result field was replaced by a formula.")
            key = f"{sheet}!{cell.coordinate}"
            group = chunks.get(key)
            if group is None:
                if isinstance(cell.value, str) and cell.value.startswith("[Full text in Text chunks: "):
                    raise ValueError("Missing workbook text chunks for a referenced cell.")
                return cell.value
            expected = f"[Full text in Text chunks: {key}; {group['total']} parts]"
            if cell.value != expected:
                raise ValueError("Text-chunk reference does not match its source cell.")
            return group["value"]

        def key_values(sheet):
            output = {}
            for row in table(sheet)[1:]:
                key = read(row[0], sheet)
                if key in output or not isinstance(key, str):
                    raise ValueError(f"Duplicate/invalid key in {sheet}.")
                output[key] = read(row[1], sheet)
            return output

        archive = key_values("Archive metadata")
        if archive.get("archive_format") != EXCEL_ARCHIVE_FORMAT:
            raise ValueError("Unsupported Excel result format.")
        summary = key_values("Run summary")
        state = {"schema_version": 2, "run_id": summary["Run ID"], "status": summary["Status"],
            "run_fingerprint": summary["Run fingerprint"], "iteration": archive["iteration"],
            "event": {"number": summary["Event number"], "Title": summary["Event title"],
                      "description": summary["Original event description"]},
            "config": {}, "metadata": {}, "nodes": [], "attempts": []}
        for row in table("Settings")[1:]:
            section, key, value, encoding = [read(c, "Settings") for c in row[:4]]
            if section not in ("config", "metadata") or key in state[section]:
                raise ValueError("Duplicate/invalid setting in workbook.")
            if encoding == "json":
                value = json.loads(value, object_pairs_hook=unique_object)
            elif encoding == "str":
                value = "" if value is None else value
                if not isinstance(value, str):
                    raise ValueError("String setting has the wrong type.")
            elif encoding == "float":
                if isinstance(value, bool) or not isinstance(value, (int, float)) or not math.isfinite(value):
                    raise ValueError("Float setting has the wrong type.")
                value = float(value)
            elif encoding == "int":
                if type(value) is not int:
                    raise ValueError("Integer setting has the wrong type.")
            elif encoding == "bool":
                if type(value) is not bool:
                    raise ValueError("Boolean setting has the wrong type.")
            elif encoding == "NoneType":
                if value is not None:
                    raise ValueError("Null setting has the wrong type.")
            else:
                raise ValueError("Unknown workbook value encoding.")
            state[section][key] = value
        rows = table("Nodes")
        labels = [c.value for c in rows[0]]
        if len(labels) != len(set(labels)):
            raise ValueError("Duplicate node columns.")
        ordered = []
        for row in rows[1:]:
            # Manual evaluation columns are intentionally excluded from completion checks.
            needed = {"Node ID", "Parent ID", "Depth", "Child index", "Complete consequence",
                      "Prompt ID", "Accepted attempt", "Seed", "Generation order"}
            needed.update(f"Sentence {i}" for i in range(1, 6))
            values = {label: read(cell, "Nodes") for label, cell in zip(labels, row) if label in needed}
            node = {"node_id": values["Node ID"], "parent_id": values["Parent ID"],
                "depth": values["Depth"], "child_index": values["Child index"],
                "sentences": [values[f"Sentence {i}"] for i in range(1, 6)],
                "text": values["Complete consequence"], "prompt_id": values["Prompt ID"],
                "accepted_attempt": values["Accepted attempt"], "seed": values["Seed"]}
            order = values["Generation order"]
            if type(order) is not int:
                raise ValueError("Invalid generation order.")
            ordered.append((order, node))
        if sorted(i for i, _ in ordered) != list(range(1, len(ordered) + 1)):
            raise ValueError("Duplicate or missing generation order.")
        state["nodes"] = [node for _, node in sorted(ordered)]
        if len(state["nodes"]) != summary["Generated nodes"]:
            raise ValueError("Workbook node count is inconsistent.")
        if digest(_workbook_resume_payload(state)) != archive.get("resume_sha256"):
            raise ValueError("Workbook result checksum mismatch; original result fields changed.")
        for row in table("Attempts")[1:]:
            record = json.loads(read(row[7], "Attempts"), object_pairs_hook=unique_object)
            if not isinstance(record, dict):
                raise ValueError("Invalid attempt record.")
            state["attempts"].append(record)
        state["performance"] = {"schema_version": archive["performance_schema_version"],
            "summary": key_values("Performance"), "definitions": key_values("Metric definitions")}
        return state
    except (KeyError, IndexError, TypeError) as exc:
        raise ValueError("Workbook is missing or has invalid result-validation fields.") from exc
    finally:
        wb.close()
````

````python
# CELL 7 — DATA VALIDATION, BRANCH GENERATION, AND RESUME LOGIC

def unique_object(pairs):
    obj = {}
    for key, value in pairs:
        if key in obj:
            raise ValueError(f"Duplicate JSON key: {key}")
        obj[key] = value
    return obj


def load_events_bytes(raw):
    payload = json.loads(raw.decode("utf-8-sig"), object_pairs_hook=unique_object)
    if isinstance(payload, dict) and set(payload) == {"events"}:
        payload = payload["events"]
    if not isinstance(payload, list):
        raise ValueError('Event JSON must be a list, or {"events": [...]} .')
    normalized = []
    for entry in payload:
        if not isinstance(entry, dict):
            raise ValueError("Each event must be an object.")
        obj = {}
        for key, value in entry.items():
            normalized_key = key.strip().lower()
            if normalized_key in obj:
                raise ValueError("Duplicate event field after case normalization.")
            obj[normalized_key] = value
        if set(obj) != {"number", "title", "description"}:
            raise ValueError("Each event must have exactly number, Title, description (case insensitive).")
        if type(obj["number"]) is not int:
            raise ValueError("Event numbers must be JSON integers, not strings or booleans.")
        for field in ("title", "description"):
            if not isinstance(obj[field], str) or not obj[field].strip():
                raise ValueError(f"Event {obj['number']} has an empty or invalid {field}.")
        words = len(obj["description"].split())
        if STRICT_EVENT_WORD_COUNT and not MIN_EVENT_WORDS <= words <= MAX_EVENT_WORDS:
            raise ValueError(f"Event {obj['number']}: {words} words; expected "
                             f"{MIN_EVENT_WORDS}–{MAX_EVENT_WORDS}.")
        normalized.append({"number": obj["number"], "Title": obj["title"].strip(),
                           "description": obj["description"].strip()})
    numbers = [e["number"] for e in normalized]
    if sorted(numbers) != list(range(1, TOTAL_EVENTS + 1)):
        raise ValueError(f"Dataset must contain each number 1–{TOTAL_EVENTS} exactly once.")
    return sorted(normalized, key=lambda e: e["number"])


def fetch_events(store):
    import requests
    if EVENTS_URL:
        parsed = urllib.parse.urlparse(EVENTS_URL)
        if parsed.scheme != "https" or parsed.username or parsed.password:
            raise ValueError("EVENTS_URL must be a credential-free HTTPS URL.")
        url = EVENTS_URL
        if parsed.netloc == "github.com" and "/blob/" in parsed.path:
            url = "https://raw.githubusercontent.com" + parsed.path.replace("/blob/", "/", 1)
        # No GitHub credentials are sent to an arbitrary event URL.
        response = requests.get(url, timeout=90)
        response.raise_for_status()
        raw = response.content
        source = {"url": url}
    else:
        dataset_commit = store.resolve_ref()
        raw = store.read_bytes(EVENTS_PATH, ref=dataset_commit)
        if raw is None:
            raise FileNotFoundError(f"Upload your event JSON to {EVENTS_PATH!r} in the repo, "
                                    "or change EVENTS_PATH in Cell 1. The repository must contain this file first.")
        source = {"repo": store.repo, "branch": GITHUB_BRANCH, "path": EVENTS_PATH, "commit": dataset_commit}
    events = load_events_bytes(raw)
    source.update({"raw_sha256": hashlib.sha256(raw).hexdigest(),
                   "normalized_sha256": digest(events), "retrieved_at": utc_now()})
    return events, source


def protocol_config(model_id):
    spec = copy.deepcopy(MODELS[model_id])
    return {
        "protocol_version": PROTOCOL_VERSION, "model_id": model_id, "model": spec,
        "num_layers": NUM_LAYERS, "children_per_node": CHILDREN_PER_NODE,
        "sentences_per_node": SENTENCES_PER_NODE, "forecast_horizon_days": FORECAST_HORIZON_DAYS,
        "base_seed": BASE_SEED, "seed_method": "sha256-v1",
        "context_tokens": CONTEXT_TOKENS, "max_new_tokens": MAX_NEW_TOKENS,
        "max_format_attempts": MAX_FORMAT_ATTEMPTS, "thinking_enabled": True,
        "thinking_budget_tokens": THINKING_BUDGET_TOKENS,
        "sampling": {**COMMON_SAMPLING, **spec["sampling"]},
        "llama_cpp_commit": LLAMA_CPP_COMMIT, "gpu_layers": GPU_LAYERS,
        "kv_cache_type": "f16", "flash_attention": True,
        "batch_size": BATCH_SIZE, "microbatch_size": MICROBATCH_SIZE,
        "prompt_cache": False, "context_shift": False,
        "system_prompt": SYSTEM_PROMPT, "user_prompt_template": USER_PROMPT_TEMPLATE,
        "sentence_validation": "five strings; full-stop ending; no internal sentence-terminal punctuation except decimals",
        "output_schema_policy": "sentences-object-array-or-single-paragraph-v3",
    }


def run_identity(model_id, event, iteration, corpus_hash):
    config = protocol_config(model_id)
    run_id = f"{MODELS[model_id]['name']}_event_{event['number']:02d}_iteration_{iteration:02d}"
    fingerprint = digest({"config": config, "event": event, "iteration": iteration,
                          "corpus_hash": corpus_hash})
    return run_id, fingerprint, config


def node_order(node_id):
    return () if node_id == "root" else tuple(map(int, node_id.split(".")))


def branch_nodes(nodes, parent_id):
    by_id = {n["node_id"]: n for n in nodes}
    result = []
    while parent_id != "root":
        node = by_id[parent_id]
        result.append(node)
        parent_id = node["parent_id"]
    return list(reversed(result))


def make_messages(event, nodes, parent_id, layer):
    branch = branch_nodes(nodes, parent_id)
    history = "\n\n".join(f"Layer {n['depth']}: {n['text']}" for n in branch)
    example = {"consequences": [{"sentences": [f"Complete sentence {i + 1}."
                for i in range(SENTENCES_PER_NODE)]} for _ in range(CHILDREN_PER_NODE)]}
    user = USER_PROMPT_TEMPLATE.format(
        event_number=event["number"], event_title=event["Title"],
        event_description=event["description"],
        branch_history=history or "None. Start from the original event.",
        children=CHILDREN_PER_NODE, layer=layer, layers=NUM_LAYERS,
        horizon_days=FORECAST_HORIZON_DAYS, sentences=SENTENCES_PER_NODE,
        json_example=json.dumps(example, ensure_ascii=False))
    return [{"role": "system", "content": SYSTEM_PROMPT}, {"role": "user", "content": user}]


def normalize_consequence_container(child, sentence_count):
    """Normalize containers; split only a single string with exactly N strict sentences."""
    if isinstance(child, list):
        child = {"sentences": child}
    if not isinstance(child, dict) or set(child) != {"sentences"}:
        return child
    sentences = child["sentences"]
    if (isinstance(sentences, list) and len(sentences) == 1
            and isinstance(sentences[0], str)
            and "\n" not in sentences[0] and "\r" not in sentences[0]):
        # Decimal points have no following whitespace, so 3.5 stays intact.
        # The validator below still checks every segment and rejects abbreviations.
        parts = re.split(r"(?<=\.)[ \t]+(?=\S)", sentences[0].strip())
        if len(parts) == sentence_count:
            return {"sentences": parts}
    return child


def validate_answer(content, finish_reason, children_count=None, sentence_count=None):
    children_count = CHILDREN_PER_NODE if children_count is None else children_count
    sentence_count = SENTENCES_PER_NODE if sentence_count is None else sentence_count
    errors = []
    if finish_reason != "stop":
        return None, [f"Expected stop, got {finish_reason!r}; truncated or incomplete output is rejected."]
    text = content.strip()
    # Strip formatting fences only; never repair, rewrite, truncate, or pad predictions.
    if text.startswith("```") and text.endswith("```"):
        text = re.sub(r"^```(?:json)?\s*", "", text)[:-3].strip()
    try:
        obj = json.loads(text, object_pairs_hook=unique_object)
    except (json.JSONDecodeError, ValueError) as exc:
        return None, ["Invalid JSON: " + str(exc)]
    if not isinstance(obj, dict) or set(obj) != {"consequences"}:
        return None, ["Expected exactly one top-level key: consequences."]
    children = obj["consequences"]
    if not isinstance(children, list) or len(children) != children_count:
        return None, [f"Expected exactly {children_count} consequences."]
    # Preserve wording, punctuation and order. Raw output remains in the attempt log.
    children = [normalize_consequence_container(child, sentence_count) for child in children]
    seen = set()
    for i, child in enumerate(children, 1):
        if not isinstance(child, dict) or set(child) != {"sentences"}:
            errors.append(f"Child {i}: expected a sentence list or an object containing only sentences.")
            continue
        sentences = child["sentences"]
        if not isinstance(sentences, list) or len(sentences) != sentence_count:
            errors.append(f"Child {i}: expected {sentence_count} sentence strings.")
            continue
        for j, sentence in enumerate(sentences, 1):
            if not isinstance(sentence, str) or not sentence.strip():
                errors.append(f"Child {i}, sentence {j}: empty/non-string.")
                continue
            s = sentence.strip()
            if "\n" in s or "\r" in s or not s.endswith(".") or len(s.split()) < 3:
                errors.append(f"Child {i}, sentence {j}: use one complete sentence ending in a full stop.")
            interior = re.sub(r"(?<=\d)\.(?=\d)", "", s[:-1])
            if any(char in interior for char in ".!?"):
                errors.append(f"Child {i}, sentence {j}: multiple sentences/dotted abbreviation/ellipsis.")
        if all(isinstance(s, str) for s in sentences):
            combined = " ".join(s.strip() for s in sentences).casefold()
            if combined in seen:
                errors.append("The two consequences are exact duplicates.")
            seen.add(combined)
    return (children if not errors else None), errors


def generation_seed(model_id, event_number, iteration, parent_id, attempt):
    material = [BASE_SEED, model_id, event_number, iteration, parent_id, attempt]
    return int(hashlib.sha256(canonical_bytes(material)).hexdigest()[:8], 16) % (2**31 - 1)


def verify_tree(state, require_complete=False):
    nodes = state["nodes"]
    b = state["config"]["children_per_node"]
    d = state["config"]["num_layers"]
    by_id = {n["node_id"]: n for n in nodes}
    if len(by_id) != len(nodes):
        raise ValueError("Checkpoint has duplicate node IDs.")
    groups = {}
    for node in nodes:
        path = node_order(node["node_id"])
        if len(path) != node["depth"] or not 1 <= len(path) <= d or any(i < 1 or i > b for i in path):
            raise ValueError("Checkpoint has an invalid node path/depth.")
        if node["child_index"] != path[-1]:
            raise ValueError("Checkpoint has an inconsistent child index.")
        parent = ".".join(map(str, path[:-1])) or "root"
        if node["parent_id"] != parent or (parent != "root" and parent not in by_id):
            raise ValueError("Checkpoint has an invalid or missing parent.")
        if len(node["sentences"]) != 5 or node["text"] != " ".join(node["sentences"]):
            raise ValueError("Checkpoint sentence/text inconsistency.")
        groups.setdefault(parent, []).append(node)
    for siblings in groups.values():
        if sorted(n["child_index"] for n in siblings) != list(range(1, b + 1)):
            raise ValueError("Checkpoint has an incomplete/invalid sibling group.")
        _, errors = validate_answer(json.dumps({"consequences": [
            {"sentences": n["sentences"]} for n in siblings]}), "stop",
            children_count=b, sentence_count=state["config"]["sentences_per_node"])
        if errors:
            raise ValueError("Checkpoint contains invalid sentences: " + "; ".join(errors))
    if require_complete and len(nodes) != sum(b**i for i in range(1, d + 1)):
        raise ValueError("Result does not contain a complete tree.")


def generate_tree(state, backend, checkpoint_path, progress=None):
    verify_tree(state)
    monitor = RunPerformanceMonitor(state)
    try:
        monitor.start()
        return _generate_tree_measured(state, backend, checkpoint_path, progress, monitor)
    finally:
        monitor.stop()
        state["performance"] = performance_report(state)
        atomic_json(checkpoint_path, state)


def _generate_tree_measured(state, backend, checkpoint_path, progress, monitor):
    """Breadth-first expansion; never uses factual accuracy to prune a branch."""
    def save_checkpoint():
        monitor.sync()
        atomic_json(checkpoint_path, state)

    verify_tree(state)
    nodes = state["nodes"]
    state["status"] = "running"
    expanded = {n["parent_id"] for n in nodes}
    if progress:
        progress.update(len(expanded))
    for depth in range(1, NUM_LAYERS + 1):
        parents = ["root"] if depth == 1 else [n["node_id"] for n in sorted(
            nodes, key=lambda n: node_order(n["node_id"])) if n["depth"] == depth - 1]
        for parent_id in parents:
            if parent_id in expanded:
                continue
            prompt_id = f"layer_{depth}_parent_{parent_id}"
            messages = make_messages(state["event"], nodes, parent_id, depth)
            # Persist attempt intent before inference. A terminated request remains 'interrupted'.
            previous = [a for a in state["attempts"] if a["prompt_id"] == prompt_id]
            attempts_used = sum(a.get("status") in ("accepted", "rejected") for a in previous)
            success = False
            for logical_attempt in range(attempts_used + 1, MAX_FORMAT_ATTEMPTS + 1):
                seed = generation_seed(state["config"]["model_id"], state["event"]["number"],
                                       state["iteration"], parent_id, logical_attempt)
                record = {"prompt_id": prompt_id, "parent_id": parent_id, "layer": depth,
                          "attempt": logical_attempt, "seed": seed, "started_at": utc_now(),
                          "performance_segment_id": monitor.segment["segment_id"],
                          "messages": messages, "status": "interrupted", "validation_errors": []}
                state["attempts"].append(record)
                save_checkpoint()
                try:
                    response = monitor.complete(backend, record, messages, seed, state["config"]["sampling"])
                except BaseException as exc:
                    record["error_type"] = type(exc).__name__
                    state["status"] = "interrupted"
                    save_checkpoint()
                    raise
                record.update(response)
                children, errors = validate_answer(response["raw_content"], response["finish_reason"])
                record["validation_errors"] = errors
                record["finished_at"] = utc_now()
                record["status"] = "rejected" if errors else "accepted"
                if errors:
                    save_checkpoint()
                    print(f"Format rejection: {prompt_id}, attempt {logical_attempt}: " + "; ".join(errors), flush=True)
                    if progress:
                        progress.set_postfix_str(f"{prompt_id}: format retry {logical_attempt}")
                    continue
                for child_index, child in enumerate(children, 1):
                    sentences = [s.strip() for s in child["sentences"]]
                    node_id = ("" if parent_id == "root" else parent_id + ".") + str(child_index)
                    nodes.append({"node_id": node_id, "parent_id": parent_id, "depth": depth,
                                  "child_index": child_index, "sentences": sentences,
                                  "text": " ".join(sentences), "prompt_id": prompt_id,
                                  "accepted_attempt": logical_attempt, "seed": seed})
                expanded.add(parent_id)
                state["updated_at"] = utc_now()
                save_checkpoint()
                if progress:
                    progress.update(1)
                    progress.set_postfix_str(f"layer {depth}/{NUM_LAYERS}")
                success = True
                break
            if not success:
                state["status"] = "format_failed"
                save_checkpoint()
                last_errors = next((a.get("validation_errors", []) for a in reversed(state["attempts"])
                                    if a["prompt_id"] == prompt_id), [])
                raise RuntimeError(f"{state['run_id']} {prompt_id}: no valid answer after "
                                   f"{MAX_FORMAT_ATTEMPTS} attempts. Saved audit data at {checkpoint_path}. "
                                   f"Last rejection: {'; '.join(last_errors)}. "
                                   "Review the pilot and protocol; this run is not uploaded as complete.")
    verify_tree(state, require_complete=True)
    state["status"] = "complete"
    state["completed_at"] = utc_now()
    save_checkpoint()
    return state


def load_completed_zip(raw, expected_id, fingerprint):
    with zipfile.ZipFile(io.BytesIO(raw)) as archive:
        names = [n for n in archive.namelist() if n == "state.json" or n.endswith("/state.json")]
        if len(names) == 1:
            # Backward compatibility: already-uploaded JSON/image ZIPs still skip safely.
            entry = archive.getinfo(names[0])
            if entry.file_size > 200 * 1024**2:
                raise ValueError("Existing state.json exceeds the permitted size.")
            state = json.loads(archive.read(entry))
        elif not names and archive.namelist() == ["results.xlsx"]:
            entry = archive.getinfo("results.xlsx")
            if entry.file_size > 100 * 1024**2:
                raise ValueError("Existing workbook exceeds the permitted size.")
            state = read_completed_workbook(archive.read(entry))
        else:
            raise ValueError("Existing ZIP has no unique legacy state or Excel-only result.")
    if state.get("run_id") != expected_id or state.get("run_fingerprint") != fingerprint:
        raise ValueError(f"{expected_id} already exists with different data/settings. "
                         "Use a new RESULTS_FOLDER for a different experiment; existing results are protected.")
    if state.get("status") != "complete":
        raise ValueError("Remote ZIP is not a completed result.")
    verify_tree(state, require_complete=True)
    return state


def run_experiment():
    from tqdm.auto import tqdm
    check_settings()
    repo = get_repo_name(GITHUB_REPO_URL)
    token = read_secret(GITHUB_TOKEN_SECRET, required=True)
    store = GitHubStore(repo=repo, branch=GITHUB_BRANCH, results_dir=RESULTS_FOLDER, token=token)
    del token
    events, source = fetch_events(store)  # Fail early before large downloads.
    store.list_dir(RESULTS_FOLDER)  # Detect a plain file named Results before inference.
    checkpoint_root = WORK_DIR
    if USE_GOOGLE_DRIVE_CHECKPOINTS:
        from google.colab import drive
        drive.mount("/content/drive")
        checkpoint_root = DRIVE_WORK_DIR
    WORK_DIR.mkdir(parents=True, exist_ok=True)
    checkpoint_root.mkdir(parents=True, exist_ok=True)
    atomic_json(checkpoint_root / "events_downloaded.json", events)
    selected_events = [e for e in events if START_EVENT <= e["number"] <= END_EVENT]
    repetitions = list(range(START_ITERATION, ITERATIONS_PER_EVENT + 1))
    prompts_per_run = sum(CHILDREN_PER_NODE**i for i in range(NUM_LAYERS))
    nodes_per_run = sum(CHILDREN_PER_NODE**i for i in range(1, NUM_LAYERS + 1))
    total = len(SELECTED_MODELS) * len(selected_events) * len(repetitions)
    print(f"Selected: {total} runs; each has {prompts_per_run} generation requests, "
          f"{nodes_per_run} nodes, and {CHILDREN_PER_NODE**NUM_LAYERS} leaf branches.")
    print("Completed GitHub ZIPs will be validated and skipped. Upload failure stops the experiment.")
    binary = None
    hf_token = None
    completed = skipped = 0
    with tqdm(total=total, desc="All selected runs", unit="run", position=0) as overall:
        for model_id in SELECTED_MODELS:
            backend = None
            model_path = None
            try:
                for event in tqdm(selected_events, desc=MODELS[model_id]["name"] + " events",
                                  unit="event", position=1, leave=False):
                    for iteration in tqdm(repetitions, desc=f"Event {event['number']} iterations",
                                          unit="run", position=2, leave=False):
                        run_id, fingerprint, config = run_identity(model_id, event, iteration,
                                                                 source["normalized_sha256"])
                        remote_path = RESULTS_FOLDER.strip("/") + "/" + run_id + ".zip"
                        remote = store.read_bytes(remote_path)
                        if remote is not None:
                            load_completed_zip(remote, run_id, fingerprint)
                            skipped += 1
                            overall.update(1)
                            overall.set_postfix(completed=completed, skipped=skipped)
                            continue
                        # Configuration-specific local directory protects interrupted alternative pilots.
                        local_dir = checkpoint_root / "runs" / (run_id + "_" + fingerprint[:12])
                        checkpoint = local_dir / "checkpoint.json"
                        if checkpoint.exists():
                            state = json.loads(checkpoint.read_text(encoding="utf-8"))
                            if state.get("run_fingerprint") != fingerprint:
                                raise ValueError("Local checkpoint fingerprint mismatch.")
                            verify_tree(state)
                        else:
                            state = {"schema_version": 1, "run_id": run_id, "run_fingerprint": fingerprint,
                                     "status": "running", "event": event, "iteration": iteration,
                                     "config": config, "nodes": [], "attempts": [],
                                     "created_at": utc_now(), "metadata": {"event_source": source,
                                     "runtime_sessions": []}}
                        if state["status"] != "complete":
                            if backend is None:
                                if binary is None:
                                    binary = build_llama_backend(WORK_DIR)
                                    hf_token = read_secret(HF_TOKEN_SECRET)
                                model_path = download_model(model_id, hf_token)
                                backend = LlamaServer(binary, model_path, MODELS[model_id]["name"], WORK_DIR)
                            state["metadata"]["runtime_sessions"].append({
                                "resumed_at": utc_now(), "environment": runtime_metadata(),
                                "backend_artifact": copy.deepcopy(globals().get("_BACKEND_METADATA", {})),
                                "attempt_start_index": len(state["attempts"]),
                                "node_count_at_start": len(state["nodes"]),
                                "server_command": backend.command,
                                "server_props": backend.props})
                            with tqdm(total=prompts_per_run, desc=run_id, unit="prompt",
                                      position=3, leave=False) as node_bar:
                                generate_tree(state, backend, checkpoint, node_bar)
                        verify_tree(state, require_complete=True)
                        archive_path = local_dir / (run_id + ".zip")
                        if not archive_path.exists() or not archive_has_performance(archive_path):
                            archive_path = export_run(state, local_dir)
                        # Read-back locally ensures an incomplete/stale export is never marked complete.
                        archive_bytes = archive_path.read_bytes()
                        load_completed_zip(archive_bytes, run_id, fingerprint)
                        print(f"Uploading {run_id}.zip...")
                        commit = store.put_bytes_new(remote_path, archive_bytes,
                                                     f"Save thesis result {run_id}")
                        atomic_json(local_dir / "upload_receipt.json", {
                            "path": remote_path, "commit": commit, "uploaded_at": utc_now(),
                            "zip_sha256": hashlib.sha256(archive_bytes).hexdigest(),
                            "run_fingerprint": fingerprint})
                        completed += 1
                        overall.update(1)
                        overall.set_postfix(completed=completed, skipped=skipped)
            finally:
                if backend:
                    backend.close()
                if DELETE_MODEL_AFTER_USE and model_path is not None:
                    shutil.rmtree(model_path.parent)
    print(f"Finished: {completed} uploaded, {skipped} already complete. "
          f"https://github.com/{repo}/tree/{GITHUB_BRANCH}/{RESULTS_FOLDER}")
````

````python
# CELL 8 — RUN OR RESUME THE EXPERIMENT
run_experiment()
````
